```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("datasets/dp_lms_student_analytics_dataset.csv")
df.head(2)
```

<h1>Line Chart</h1>

```python
df["EnrollmentDate"] = pd.to_datetime(df["EnrollmentDate"])

monthly_count = df.groupby(df["EnrollmentDate"].dt.to_period("M")).size()
monthly_count.index = monthly_count.index.to_timestamp()

plt.figure(figsize=(10,5))
plt.plot(monthly_count.index, monthly_count.values)
plt.title("Monthly Student Enrollment")
plt.xlabel("Month")
plt.ylabel("Number of Students")
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
```

```python
monthly_revenue = df.groupby(df["EnrollmentDate"].dt.to_period("M"))["Revenue"].sum()
monthly_revenue.index = monthly_revenue.index.to_timestamp()
plt.figure(figsize= (10,12))
plt.plot(monthly_revenue.index, monthly_revenue.values)
plt.title("Monthly Revenue Trends")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
```

```python
course_progress = df.groupby( df["EnrollmentDate"].dt.to_period("M"))["CourseProgress"].mean()
course_progress.index = course_progress.index.to_timestamp()

plt.figure(figsize= (10,12))
plt.plot(course_progress.index, course_progress.values)
plt.title("Average Course Progress Per Month")
plt.xlabel("Month")
plt.ylabel("Progress")
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
```

```python
attendance_percent = df.groupby(df["EnrollmentDate"].dt.to_period("M"))["AttendancePercentage"].mean()
attendance_percent.index = attendance_percent.index.to_timestamp()

plt.figure(figsize= (10,12))
plt.plot(attendance_percent.index , attendance_percent.values)
plt.title("Average Attendance by Month")
plt.xlabel("Month")
plt.ylabel("Attendance")
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
```

<h2> Bar Charts</h2>

```python
course_counts = df['Course'].value_counts()

plt.figure(figsize=(10, 6))
plt.bar(course_counts.index, course_counts.values, color='skyblue')
plt.title('Number of Students in Each Course')
plt.xlabel('Course')
plt.ylabel('Number of Students')
plt.xticks(rotation=45)

plt.tight_layout()

```

```python
course_revenue = df.groupby('Course')['Revenue'].sum()

plt.figure(figsize=(10, 6))
plt.bar(course_revenue.index , course_revenue.values)
plt.title('Revenue Generated by each Course')
plt.xlabel('Course')
plt.ylabel('Revenue')
plt.xticks(rotation=45)

plt.tight_layout()
```

```python
avg_final_score = df.groupby('Course')['FinalScore'].mean()

plt.figure(figsize=(10,6))
plt.bar(avg_final_score.index, avg_final_score.values , color='orange', edgecolor='black')
plt.title('Average Final Score by Course')
plt.xlabel('Course')
plt.ylabel('Average Final Score')

plt.xticks(rotation=45)
```

```python
avg_attendance = df.groupby('Course')['AttendancePercentage'].mean()

plt.bar(avg_attendance.index, avg_attendance.values, color='purple', edgecolor='black')
plt.title('Average Attendance Percentage by Course')
plt.xlabel('Course')
plt.ylabel('Average Attendance (%)')

plt.xticks(rotation=45)
```

```python
city_students = df['City'].value_counts()

plt.bar(city_students.index, city_students.values, color='green', edgecolor='black')
plt.title('Number of Students From Each City')
plt.xlabel('City')
plt.ylabel('Number of Students')

plt.xticks(rotation=45)
```

```python
instructor_students = df['Instructor'].value_counts()

plt.figure(figsize=(10,6))
plt.bar(instructor_students.index, instructor_students.values, color='red', edgecolor='black')
plt.title('Number of Students Under Each Instructor')
plt.xlabel('Instructor')
plt.ylabel('Number of Students')

plt.xticks(rotation=45)
```

```python
course_enrollment = df['Course'].value_counts()
top_10_courses = course_enrollment.head(10)

plt.figure(figsize=(10,6))
plt.barh(top_10_courses.index, top_10_courses.values, color='skyblue', edgecolor='black')

plt.title('Top 10 Courses by Enrollment')
plt.xlabel('Number of Students')
plt.ylabel('Course')
plt.tight_layout()
```

```python
instructor_enrollment = df['Instructor'].value_counts()
top_10_instructors = instructor_enrollment.head(10)

plt.figure(figsize=(10,6))
plt.barh(top_10_instructors.index, top_10_instructors.values, color='yellow', edgecolor='black')

plt.title('Top 10 Instructors by Student Enrollment')
plt.xlabel('Number of Students')
plt.ylabel('Instructor')
```

```python
high_score= df.sort_values(by='FinalScore', ascending=False).head(10)

plt.figure(figsize=(10,6))
plt.barh(high_score["StudentName"], high_score["FinalScore"], color='orange', edgecolor='black')

plt.title('Top 10 Students with Highest Final Score')
plt.xlabel('Final Score')
plt.ylabel('Student Name')
```
